### Dropping Duplicates

```python
def drop_duplicates(df):
    return df.drop_duplicates()
drop_duplicates(df)
```

##### -> The above function is removing all duplicates value at one time

### Handling Missing Values

```python
def handle_missing(df):
    for col in df.columns:
        if df[col].dtype in ['int64','float64']: #selecting only numerical columns.
            df[col].fillna(df[col].median(), inplace=True)
        else:
            df[col].fillna(df[col].mode()[0], inplace=True)
    return df

handle_missing(df)
```

##### -> The above function is filling all NaN values of numerical columns and categorical columns with their median and mode respectively

### Handling Outliers

```python
def handle_outliers(df, num_cols):
    for col in num_cols:
        Q1 = df[col].quantile(0.25)
        Q3 = df[col].quantile(0.75)
        IQR = Q3 - Q1
        lower = Q1 - 1.5*IQR
        upper = Q3 + 1.5*IQR
        df[col] = df[col].clip(lower, upper)
    return df

num_cols=['release_year','duration_num']
handle_outliers(df, num_cols)
```

## Function Explained
- 
- Q1 is representing 25th percentile means lower quartile.
- Q3 is representing 75th percentile means upper quartile.
- IQR = Q3-Q1 computes 50% of the data means measures the spread of the middle.
- lower = Q1 - 1.5*IQR and upper = Q3 + 1.5*IQR is ensuring if anything is out of this range, then it is considered as a outliers.
- clip(lower, upper) func replaces values below lower with lower, and values above upper with upper, so instead of removing outliers we can make boundary.

### Numerical Column Plots

```python
df.select_dtypes(include=['int64', 'float64']).columns
```

```python
def plot_numerical(df, num_cols):
    for col in num_cols:
        fig, ax = plt.subplots(1, 2, figsize=(12,4))
        sns.histplot(df[col], kde=True, ax=ax[0])
        sns.boxplot(x=df[col], ax=ax[1])
        fig.suptitle(col)  
        plt.show()

num_cols=['release_year', 'duration_num']
plot_numerical(df, num_cols)
```

## Function Explained
- fig is creating figure and 'ax' is creating a grid of subplots.
- ax=1,ax=2 means one row and two columns, so you get side by side two plots..
- suptitle func shows title for respective column, so that there is no confusion.

### Categorical Column Plots

```python
df.select_dtypes(include=['object']).columns
```

```python
df.drop(columns=['show_id','date_added','duration'],inplace=True)
```

```python
top_countries = df['country'].value_counts().head(10).index
df['country'] = df['country'].where(df['country'].isin(top_countries), 'Other')
```

```python
top_genres = df['listed_in'].value_counts().head(10).index
df['listed_in'] = df['listed_in'].where(df['listed_in'].isin(top_genres), 'Other')
```

## Observation 
- As country and listed_in columns have a large amount of data so it is selecting only main data which represents majority
- where(condition, other) func keeps the original value where the condition is True otherwise if False, replace it with 'Other'

### Observation
#### Categorical Feature Selection
- Identifier and date-based columns were excluded from categorical analysis
- Only business-relevant categorical features were used for visualization

```python
def plot_categorical(df, cat_cols):
    for col in cat_cols:
        plt.figure(figsize=(6,4))
        sns.countplot(x=df[col])
        plt.title(col)
        plt.xticks(rotation=45) #rotates the country name 45% so that it should be visible properly.
        plt.show()

cat_cols=['country', 'rating','listed_in']
plot_categorical(df, cat_cols)
```

## Observation
1. Production of Movies is approximately twice of TV shows which shows that viewers like movies more over TV shows
2. USA has the highest number of releases followed by India
3. TV-MA is the most common rating on Netflix which shows that Netflix content is targeted Mature and teenage audience more than Family friendly
4. Drama related genres dominate Netflix library with International Movies and Comedies while Kids and family movies has small proportion of catalog

## Target vs Feature

```python
df['is_movie'] = (df['type'] == 'Movie').astype(int)
```

### Observation 
We have to make is_movie numerical column as it is required to represent target
- 1 → Movie
- 0 → TV Show

```python
def plot_cat_vs_target(df, cat_cols, target):
    for col in cat_cols:
        sns.barplot(x=col, y=target, data=df)
        plt.title(f"{col} vs {target}")
        plt.xticks(rotation=45)
        plt.show()

cat_cols = ['country', 'rating','listed_in']
plot_cat_vs_target(df, cat_cols,'is_movie')
```
